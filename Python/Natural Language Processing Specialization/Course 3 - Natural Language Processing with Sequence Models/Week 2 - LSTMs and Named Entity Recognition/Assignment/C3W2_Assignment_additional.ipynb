```python
import tensorflow as tf
import numpy as np
```

```python
y_pred = [[0.1,0.6,0.3] , [0.2,0.7,0.1], [0.1, 0.5,0.4], [0.4,0.4,0.2]]
y_pred
```

```python
y_pred_class = tf.math.argmax(y_pred, axis=-1)
y_pred_class
```

```python
y_pred_class = tf.cast(y_pred_class, tf.float32) 
```

```python
y_pred_class
```

```python
y_true = [0,1,2,0]
```

```python
y_true = tf.cast(y_true, tf.float32) 
y_true
```

```python
matches_true_pred  = tf.equal(y_true, y_pred_class)
matches_true_pred
```

```python
matches_true_pred = tf.cast(matches_true_pred , tf.float32) 
matches_true_pred
```

```python
true_labels_1 = [0,-1,2,0]
true_labels_1 = tf.cast(true_labels_1, tf.float32)
mask = tf.math.not_equal(true_labels_1, -1)
mask
```

```python
mask = tf.cast(mask, tf.float32) 
mask
```

```python
matches_true_pred *= mask
matches_true_pred
```

```python
masked_acc = tf.reduce_sum(matches_true_pred) / tf.reduce_sum(mask)
masked_acc
```
